```python
# alpha example
import matplotlib.pyplot
import numpy as np

X = data = np.random.random((100,100))
ax_img = matplotlib.pyplot.imshow(X)
ax_img.set_alpha(0.3)
```

```python
# clim example
import matplotlib.pyplot
import numpy as np

X = data = np.random.random((100,100))
ax_img = matplotlib.pyplot.imshow(X)
ax_img.set_clim(vmin=0.2, vmax=0.5)
```

```python
# clip_box example
import matplotlib.pyplot as plt
import matplotlib.transforms
import numpy as np

X = data = np.random.random((5,2))
ax_img = matplotlib.pyplot.imshow(X,extent=(0,1,0,2.5))
plt.axis([0, 1, 0, 1])

ax = plt.gca()
bbox = matplotlib.transforms.TransformedBbox(matplotlib.transforms.Bbox([[0, 1.7], [1, 0]]), ax.transAxes)
ax_img.set_clip_box(bbox)
```

```python
# clip_on example
import matplotlib.pyplot as plt
import matplotlib.transforms
import numpy as np

X = data = np.random.random((5,2))
ax_img = matplotlib.pyplot.imshow(X,extent=(0,1,0,2.5))
plt.axis([0, 1, 0, 1])
ax_img.set_clip_on(False)
```
