```python
import traitlets
import matplotlib.pyplot as plt
traitlets.link
ax = plt.gca()

# traitlets.link((ax, 'xlim'), (rangexy, 'xlim'))
 
class RangeXY(traitlets.HasTraits):
    xlim = traitlets.Tuple()
    ylim = traitlets.Tuple()
    def __init__(self, ax):
        self.ax = ax
        self.ax.callbacks.connect('xlim_changed', self.update_xlim)
        self.ax.callbacks.connect('ylim_changed', self.update_ylim)
    def update_xlim(self, event):
        print(event)
        self.xlim = self.ax.get_xlim()
        print(event)
    def update_ylim(self, event):
        self.ylim = self.ax.get_ylim()
    def update_ax(self, change):
        self.ax.set_xlim(self.xlim)
        self.ax.set_ylim(self.ylim)
    def link(self, other):
        self.observe(other.update_ax, ['xlim', 'ylim'])
        other.observe(self.update_ax, ['xlim', 'ylim'])

rangexy = RangeXY(ax)
```
